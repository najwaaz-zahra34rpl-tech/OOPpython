Atribut hp dalam kode ini bersifat publik, sehingga nilainya bisa diakses dan diubah secara langsung dari luar kelas.

```python
class Hero:
    # Constructor: Dijalankan saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
        self.name = name  # Nama Hero
        self.hp = hp  # Nyawa (Health Point)
        self.attack_power = attack_power  # Kekuatan Serangan

    # Method untuk menampilkan info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method diserang: Menerima damage
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")


# ==========================================
# -- Main Program --
# ==========================================

# Membuat Object (Instansiasi)
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# Memanggil Method Info
hero1.info()
hero2.info()

# Pertarungan
print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2)  # Layla menyerang Zilong
hero2.serang(hero1)  # Zilong membalas
```

Parameter lawan perlu menerima objek utuh (bukan sekadar string nama) karena metode serang harus memodifikasi status internal dari hero yang diserang.

Jika hanya mengirimkan string nama (misalnya "Zilong"), program tidak akan tahu objek mana di memori yang harus dikurangi HP-nya.

```python
class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")


# Class Mage adalah anak dari class Hero (Inheritance)
class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)  # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")


# -- Main Program Baru --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond)  # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond)  # Skill khusus Mage
```

```python
class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di dalam class ini)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0  # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        # Kita pakai setter/getter bahkan di dalam class sendiri agar aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")


# -- Uji Coba --
hero1 = Hero("Layla", 100)

# hero1.__hp = 9999 # GAGAL (Tidak akan mengubah hp asli)
# print(hero1.__hp) # ERROR (Tidak bisa dibaca langsung)

hero1.set_hp(-50)  # Coba set negatif
print(hero1.get_hp())  # Output: 0 (Karena dicegat oleh logika Setter)
```

1. Percobaan Hacking: Access Modifiers & Name Mangling
Pertanyaan: Apakah nilai HP muncul atau error? Mengapa Python mengizinkan akses ini?

Hasil: Nilai HP akan muncul (output menunjukkan 0). Tidak terjadi AttributeError.

Mengapa Python Mengizinkan Ini (Name Mangling):
Python tidak memiliki fitur private variable yang sepenuhnya terkunci secara hardware/compiler seperti pada bahasa Java atau C++. Sebagai gantinya, Python menggunakan mekanisme Name Mangling. Saat kamu menulis variabel dengan dua garis bawah (__hp), secara otomatis interpreter Python mengubah nama variabel tersebut di belakang layar menjadi _NamaClass__namaVariabel (dalam hal ini _Hero__hp).

Mengapa Tetap Tidak Boleh Dilakukan:

Merusak Kebijakan Enkapsulasi: Menembus Name Mangling memotong semua filter validasi yang ada di setter.

Konvensi dan Etika Kode (Pythonic Way): Awalan __ adalah pesan tegas dari pembuat kode kepada pengembang lain: "Data ini adalah internal class, jangan diakses langsung dari luar!".

Risiko Bug & Refactoring: Jika struktur internal class diperbarui di kemudian hari, kode luar yang memaksa akses _Hero__hp akan langsung rusak (broken code).

2. Uji Validasi: Pentingnya Setter untuk Integritas Data
Pertanyaan: Apa yang terjadi saat hero1.set_hp(-100) tanpa logika if/elif?

Hasil pada Data HP: Nilai __hp akan berubah menjadi -100.

Mengapa Method Setter Penting untuk Integritas Data:

Mencegah Data Tak Valid (Data Corruption): Dalam konteks game, HP bernilai negatif (-100) tidak masuk akal secara logika permainan. Nilai batas minimal HP seharusnya adalah 0 (mati).

Mencegah Cheating / Input Ilegal: Tanpa aturan pembatas (seperti elif nilai_baru > 1000), pemain bisa memasukkan HP 999999 dan merusak keseimbangan game (balance).

Prinsip Single Source of Truth: Logika dan batasan nilai terpusat di dalam setter. Jika aturan batas HP berubah di masa depan, kamu cukup mengubahnya di dalam method set_hp(), tanpa perlu memperbaiki kode di puluhan tempat lain yang mengubah nilai HP.
